```python
import pandas as pd
from matplotlib import pyplot as plt
```

```python


csv_path = "file_counts.csv"

df = pd.read_csv(csv_path)

count = df["file_count"]

# count strucs less than threshold 
threshold = 10
count_lt_n = (count < threshold).sum()

print(f"Percentage of samples with less than {threshold} trajectories: {count_lt_n / len(count) * 100}%")

plt.figure(figsize=(10, 6), dpi=300)
plt.hist(count, bins=50, edgecolor="black", color="steelblue")
plt.title("Distribution of Trajectory Counts per Sample")
plt.xlabel("Number of Files")
plt.ylabel("Number of Folders")
plt.tight_layout()
plt.show()
```
