```python
from ase.io import read
from ase.visualize import view
from ase.geometry.cell import Cell as AseCell

filepath = 'POSCAR.vasp'
structure = read(filepath)

# Reorient cell: a → x, b → xy-plane
cell_std, Q = AseCell(structure.get_cell()).standard_form()
structure.set_positions(structure.get_positions() @ Q.T)
structure.set_cell(cell_std)

# Wrap atoms into the unit cell
structure.wrap()

# View the structure
view(structure)

view(structure)
```
