```python
import numpy as np 
from matplotlib import pyplot as plt
```

## Velocity loss data 

```python
# distribution of atomic numbers 
loss_Z = np.load("loss-atomic_number.npy")

# Get unique values and their counts
unique, counts = np.unique(loss_Z.round(3), return_counts=True)


plt.figure(dpi=400)
plt.bar(unique, counts, edgecolor='black')
plt.xticks(unique, rotation=90)   # force each unique value to appear
plt.tight_layout()
plt.show()
```

```python
# distribution of tags 
loss_tag = np.load("loss-tag.npy")

# Get unique values and their counts
unique, counts = np.unique(loss_tag.round(3), return_counts=True)

plt.figure(dpi=400)
plt.bar(unique, counts, edgecolor='black')
plt.xticks(unique, rotation=90)   # force each unique value to appear
plt.tight_layout()
plt.show()
```

```python
# distribution of atomic distance between surface atom and adsorbate 
loss_displacement = np.load("loss-displacement_from_ads.npy")

plt.figure(dpi=400)
plt.hist(loss_displacement, bins=30, edgecolor='black')
plt.xlabel('Displacement from adsorbate')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
distance_max = np.load("distance-max.npy")
distance_max_loss = np.load("distance-max_loss.npy")

pairs = np.stack([distance_max, distance_max_loss], axis=1)

unique_pairs, counts = np.unique(pairs, axis=0, return_counts=True)

plt.figure(dpi=400)
plt.scatter(
    unique_pairs[:, 0],
    unique_pairs[:, 1],
    s=5 + counts,      # or e.g. 5 + 3*counts to exaggerate
    alpha=0.7,
)
plt.xlabel("distance_max")
plt.ylabel("distance_max_loss")

```
